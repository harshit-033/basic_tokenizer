```python


with open("words.txt",'r',encoding="utf-8") as file:
    content=file.read()

print(len(content))
```

```python
import re
```

```python
preprocessed=re.split(r'([,.:;?_!"()\']|--|\s)',content)
preprocessed=[text for text in preprocessed if text.strip() ]
```

```python
print(preprocessed)
```

```python
total=sorted(list(set(preprocessed)))
total.extend(["|<endofline>|","|<unk>|"])
vocab={t:i for t,i in enumerate(total)}
```

```python
class tokenizer:
    def __init__(self,vocab):
        self.str_to_int=vocab
        self.int_to_str={i:s for s,i in vocab.items()}

    def encoder(self,text):
        preprocessed=re.split(r'([,.:;?_!"()\']|--|\s)',text)
        preprocessed=[item.strip() for item in preprocessed if item.strip()]
        preprocessed=[
            item if item in self.str_to_int
            else "|<unk>|" for item in preprocessed
        ]
        id=[self.str_to_int[s] for s in preprocessed]
        return id
    def decoder(self,ids):
        text=" ".join([self.int_to_str[s] for s in ids])
        text=re.sub(r'\s+([,.:;?!"()\'])',r'\1',text)
        return text
```

```python
machine1=tokenizer(vocab)
```

```python
t1="hello how are you?my name is harshit"
id1=machine1.encoder(t1)
```

```python
total = sorted(list(set(preprocessed)))
total.extend(["|<endofline>|", "|<unk>|"])

vocab = {t: i for i, t in enumerate(total)}

machine1 = tokenizer(vocab)
```

```python
t1="hello how are you?my name is harshit"
id1=machine1.encoder(t1)
```

```python
print(id1)
```

```python
text=machine1.decoder(id1)
```
